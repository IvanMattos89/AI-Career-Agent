#importar bibliotecas Gemini e Crewai

```python
import os
import asyncio
import nest_asyncio
from google import genai
from google import genai
from google.colab import userdata
from crewai import Agent, Task, Crew, LLM
```

```python
#aplicar o caminho para permitir as chhamadas sincroas/
#assincronas dentro do laço (loop) de repetição do Colab
nest_asyncio.apply()
```

```python
#configurar a chave do gemini
os.environ["GOOGLE_GENAI_PROJECT"] = "crewai"
#os.environ{"GOOGLE_API_KEY"} = userdata.get(GEMINI_API_kEY)#MESMA API DOS PROJETOS ANTERIORES
```

```python
#lista os modelos disponíveis
client = genai.Client(api_key=os.environ.get("GOOGLE_GENAI_PROJECT"))

GEMINI_LLM = LLM(
    model="gemini/gemini-1.5-flash",
    api_key=userdata.get("GEMINI_API_KEY") if "GEMINI_API_KEY" in dir(userdata) else os.environ.get("GEMINI_API_KEY"),
    temperature=0.7,
    max_retries=5,
    retry_delay=10 #espera de 10s entre tentativas
)
```

```python
criador_social = Agent(
    #definir a personalidade
    role='Criador de conteúdo para Redes Sociais com mais de 15 anos de experiência',
    goal='Produzir post envolventes sobre tecnologia e inovação de mercado, considerando a contemporaniedade.',
    backstory=(
        "especialista em transformar conceitos técnicos em mensagem acessíveis e cativantes para deres sociais como Instagram , Linkedin e TikTok"
    ),
    llm=GEMINI_LLM,
    verbose=True
)
```

```python
#agente Editor de coteúdo
editor_social = Agent(
    role="Editor de COnteúdo Digital Sênior",
    goal="Ajustar o tom , clareza, legibilidade e formato dos posts para maximizar o engajamento e interação",
    backstory=("Profissional com 20 anos de experiência em conteúdo de copywriting e otimização para plataformas com LinKedin, Reddit, Stackoverflow, Instagram e TikTok"),
    llm=GEMINI_LLM,
    verbose=True
)
```

```python
#Criação da Tarefa do post para criador_social

tarefa_post=Task(
    description=(
        "Crie um post para redes socias explicando de forma simples o que são agentes autonomos de IA, e como eles podem ser utilizados pos profissionais de marketing e criadores de conteúdo. O post deve ser curto, direto, objetivo com tom inspirador."

    ),
    expected_output="Um post com até 280 caractes, pronto para publicação, com linguagem envolvente e acessível, não faça utilização de termos muito técnicos",
    agent=criador_social,
)
```

```python
#Criação da Tarefa de Edição para o Agent editor_social

tarefa_revisao=Task(
    description="Revise o conteúdo do post criado sobre agentes autonomos de IA para redes sociais. Ajuste o tom para torná-lo acessível e inspirador, corrija eventuais erros de linguagem e otimime o formato para as plataforma como linkedln e Instagram",
    expected_output="Post revisado para redes sociais, tom envolvente e pronto para revisão humana para ser publicado.",
    agent=editor_social
)
```

```python
# Criação do time/equipe dos agentes
equipe_social = Crew(
    agents=[criador_social, editor_social],
    tasks=[tarefa_post, tarefa_revisao],
    verbose=True
)
```

```python
# No Colab, podemos usar o 'await' diretamente sem precisar do asyncio.run()
async def executar_crew():
    return await equipe_social.kickoff_async()

resultado = await executar_crew()
print(resultado)
```
